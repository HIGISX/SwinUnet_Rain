```python
import netCDF4 as nc

# Replace 'your_file.nc' with the path to your NetCDF file
file_path = '/home/dc/hdd1/LowRes/Carbontracker/CT2022_flux/CT2022.flux1x1.20000101.nc'

# Open the NetCDF file
dataset = nc.Dataset(file_path, 'r')

# Print the dataset to see its contents
print(dataset)
```

```python
import numpy as np


bio = dataset.variables['bio_flux_opt'][:]
fire = dataset.variables['fire_flux_imp'][:]
fossil = dataset.variables['fossil_flux_imp'][:]
ocn = dataset.variables['ocn_flux_opt'][:]

# Convert the data to a NumPy array
bio_array = np.array(bio)
fire_array = np.array(fire)
fossil_array = np.array(fossil)
ocn_array = np.array(ocn)

print(bio_array.shape)
print(fire_array.shape)
print(fossil_array.shape)
print(ocn_array.shape)
```

```python
concatenated_array = np.stack((bio_array, fire_array, fossil_array, ocn_array), axis=-1)
print(concatenated_array.shape)
```
